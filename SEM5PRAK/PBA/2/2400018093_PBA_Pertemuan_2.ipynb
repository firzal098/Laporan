## **📖 Pengantar Materi**

**🔧 PREPROCESSING**

Preprocessing adalah langkah pertama dalam pengolahan data yang bertujuan untuk membersihkan, menyiapkan, dan menyusun data mentah. Dengan demikian, data dapat diproses, dianalisis, atau digunakan lebih mudah dalam algoritma dan model pembelajaran mesin.

🎯Tujuan Utama Preprocessing:

1. Mengurangi noise atau gangguan dalam data
2. Menstandarisasi format data
3. Meningkatkan akurasi dan efisiensi model NLP atau algoritma lainnya
4. Menyaring informasi yang tidak relevan atau redundan

Dalam Pemrograman Bahasa Alami, preprocessing digunakan untuk membersihkan dan menstrukturkan teks sehingga siap dianalisis. Teks alami biasanya tidak rapi: banyak mengandung simbol, singkatan, ejaan tidak baku, dan kata-kata yang tidak penting.

**🧹 FILTERING**

Filtering adalah proses menyaring atau memilih elemen tertentu dari teks berdasarkan kriteria tertentu, untuk menghilangkan kata atau informasi yang tidak relevan, tidak penting, atau mengganggu proses analisis.

Filtering biasanya dilakukan setelah tahap tokenization, ketika teks sudah dipecah menjadi token-token (kata, kalimat, atau karakter).

**🎯 Tujuan Filtering:**
1. Mengurangi noise dalam data teks
2. Menyimpan hanya kata-kata penting
3. Meningkatkan kinerja dan akurasi model
4. Membantu fokus pada fitur-fitur yang relevan

**⚙️ Jenis Filtering:**
1. Stopword Filtering
2. Length-based Filtering
3. Part-of-Speech (POS) Filtering
4. Frequency-based Filtering

**🔪 TOKENIZATION**

Tokenisasi adalah proses memecah teks menjadi bagian-bagian kecil yang disebut token. Token bisa berupa kata (word), kalimat (sentence), atau bahkan karakter, tergantung kebutuhan. Tokenisasi digunakan untuk mempermudah dalam menganalisis sebuah teks dan mengonversi tek menjadi bentuk yang bisa diproses oleh computer.

**Jenis Tokenisasi:**
1.	Word Tokenization: memecah kalimat menjadi kata
2.	Sentence Tokenization: memecah paragraph menjadi kalimat
3.	Character Tokenization: memecah teks menjadi karakter

### **1. DATA CLEANING**

```python
# Melakukan filtering dari 5 data yang telah dislice sebelumnya
# Filter data berdasarkan suatu kondisi (contoh: tweet yang mengandung kata “efisiensi”)
filter_df = df2[df2['full_text'].str.contains('efisiensi', case=False, na=False)]
filter_df
```

```python
# Ekstrak kolom 'full_text'
full_text_column = filter_df['full_text']

print(full_text_column.head())
```

#### 1a. Menghilangkan simbol dan karakter

```python
import re

# Membuat function untuk cleaning data
def clean_text(text):
  text = re.sub(r'http\S+', '', text)         # Remove URLs
  text = re.sub(r'@\S+', '', text)            # Remove mentions (@username)
  text = re.sub(r'#\S+', '', text)            # Remove hashtags (#hashtag)
  text = re.sub(r'[^a-zA-Z\s]', '', text)     # Remove special characters and numbers
  text = re.sub(r'\s+', ' ', text).strip()    # Remove extra whitespaces
  return text
```

```python
# contoh penggunaan
raw_sample = filter_df['full_text'].iloc[3]
cleantext = clean_text(raw_sample)

# lihat bedanya antara isi variabel raw_sample dan cleantext
print('Raw data\t: ', raw_sample)
print('Cleaning\t: ', cleantext)
```

#### 1b. Casefolding

Casefolding adalah proses mengubah semua huruf dalam teks menjadi salah satu format kasus (biasanya huruf kecil). Tujuannya adalah untuk memastikan bahwa kata-kata yang sama, meskipun ditulis dengan huruf besar atau kecil, dianggap identik selama analisis teks. Ini sangat penting dalam pemrosesan bahasa alami agar komputer tidak memperlakukan 'Kata' dan 'kata' sebagai dua entitas yang berbeda.

```python
# Membuat function untuk casefolding
def casefolding(text):
  text = text.lower()                               # Mengubah teks menjadi lower case
  return text
```

```python
# contoh penggunaan
raw_sample = filter_df['full_text'].iloc[0]
case_folding = casefolding(raw_sample)

print('Raw data\t: ', raw_sample)
print('Case folding\t: ', case_folding)
```

#### 1c. Stopwords

Stopwords adalah kata-kata yang umum dan sering muncul dalam suatu bahasa (seperti 'dan', 'yang', 'di', 'untuk', 'adalah' dalam Bahasa Indonesia) namun memiliki sedikit nilai informatif untuk analisis teks. Dalam Natural Language Processing (NLP), kata-kata ini biasanya dihapus dari teks selama tahap preprocessing untuk:

```python
# Cek jumlah stopwords berbahasa Indonesia yang disediakan NLTK
len(stopwords_ind)
```

```python
# Lihat daftar stopword yang disediakan NLTK
stopwords_ind[:20]
```

```python
# Membuat function untuk menghapus stopwords
def remove_stop_words(text):
  clean_words = []
  text = text.split()
  for word in text:
      if word not in stopwords_ind:
          clean_words.append(word)
  return " ".join(clean_words)
```

```python
raw_sample = filter_df['full_text'].iloc[0]
case_folding = casefolding(raw_sample)
stopword_removal = remove_stop_words(case_folding)

print('Raw data\t\t: ', raw_sample)
print('Case folding\t\t: ', case_folding)
print('Stopword removal\t: ', stopword_removal)
```

#### 1d. Apply semua step filtering

Mengaplikasikan tahapan cleaning, casefolding, dan stopwords ke dalam dataset

```python
# Buat function untuk mengaplikasikan seluruh step Filtering

def text_preprocessing_process(text):
  text = clean_text(text)
  text = casefolding(text)
  text = remove_stop_words(text)
  return text
```

```python
filter_df['cleaned_text'] = filter_df['full_text'].apply(text_preprocessing_process)

filter_df[['full_text', 'cleaned_text']].head()
```

```python
# Lakukan tokenisasi

import nltk
nltk.download('punkt')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize


```

#### 2a. Melakukan tokenisasi dari kolom cleaned_text

```python
# Buat function untuk tokenisasi text
def tokenize_text(text):
  tokens = word_tokenize(text)
  return tokens
```

```python
# akan ada kolom baru bernama "tokenized_text" yang berisi token hasil cleaned_text
filter_df['tokenized_text'] = filter_df['cleaned_text'].apply(tokenize_text)
print(filter_df[['cleaned_text', 'tokenized_text']].head())
```

#### 2b. Menghitung Frekuensi Token

```python
# Cek frekuensi (jumlah kemunculan) token

cek_freq = filter_df['tokenized_text'].explode()  # Pisahkan semua kata/token dari setiap baris menggunakan fungsi explode()
cek_freq = cek_freq.value_counts()                # Hitung jumlah kata/token menggunakan fungsi value_counts
cek_freq.head(10)                                 # Tampilkan 10 data teratas
```
