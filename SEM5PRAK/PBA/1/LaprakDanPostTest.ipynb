**A.	Pengenalan Python**

Python adalah bahasa pemrograman tingkat tinggi yang populer, mudah dipelajari, dan serbaguna, diciptakan oleh Guido van Rossum pada 1991. Python banyak digunakan untuk analisis data, machine learning, pengembangan web, dan otomatisasi tugas karena sintaksnya yang sederhana, bersih, serta mirip bahasa Inggris, menjadikannya ideal bagi pemula.

Keunggulan Python:
* Interpreted: Kode langsung dijalankan tanpa perlu kompilasi terpisah, mempercepat proses pengembangan.
* Open Source & Gratis: Bebas digunakan dan disebarluaskan.
* Lintas Platform: Dapat berjalan di Windows, Mac, dan Linux.
* Berorientasi Objek: Mendukung pemrograman berorientasi objek, fungsional, dan prosedural.
* Library yang Berlimpah: Library-library yang telah tersedia sangat banyak dan memudahkan tugas seorang pengembang.

Kegunaan Utama:
* Data Science & ML: Pustaka (library) seperti NumPy, Pandas, Scikit-Learn.
* Pengembangan Web: Framework Django dan Flask.
* Otomatisasi (Scripting): Mengotomatiskan tugas harian, pengolahan file.

Penulisan Syntax Python
* Case Sensitive: Artinya variabel dengan nama “PBA” berbeda dengan “Pba”.
* Variabel: Dinamis, tidak perlu dideklarasikan tipe datanya dan dapat diubah ketika program dijalankan. Karakter pertama wajib huruf atau underscore sedangkan selanjutnya boleh huruf, underscore, maupun angka.
* Komentar Cukup diawali dengan tanda pagar (#), atau jika ingin beberapa garis sekaligus gunakan tiga tanda petik (“””) di awal dan di akhir.
* Tipe Data: Python memiliki beberapa tipe data yang berbeda jika dibandingkan dengan bahasa pemrograman lainnya. Berikut beberapa tipe data pada python:
    1. Boolean: True atau False
    2. String: Menyatakan karakter/kalimat bisa berupa huruf, angka, dll (diapit di dalam tanda " atau ').
    3. Integer: Menyatakan bilangan bulat.
    4. Float: Menyatakan bilangan yang mempunyai koma.
    5. Hexadecimal:	Menyatakan bilangan dalam format heksa (bilangan berbasis 16).
    6. Complex:	Menyatakan pasangan angka real dan imajiner.
    7. List:	Data untaian yang menyimpan berbagai tipe data tapi isinya bisa diubah-ubah.
    8. Tuple:	Data untaian yang menyimpan berbagai tipe data tapi isinya tidak bisa diubah.
    9. Dictionary:	Data untaian yang menyimpan berbagai tipe data berupa pasangan, penunjuk, dan nilai.

##### Menggunakan .loc[] dan .iloc[]


---



Properties .loc[ ] dan .iloc[ ] dari library Pandas sama-sama berfungsi untuk mengakses dan menampilkan data secara spesifik berdasarkan lokasi row dan column.

Perbedaan dasar .loc[ ] dan .iloc[ ] adalah sbb:

- .loc[ ] mengakses data berdasarkan nama kolom/label, bisa disertai kondisional (sama dengan, lebih dari, kurang dari, dll)
- .iloc[ ] mengakses data berdasarkan row/index/baris, hanya bisa diisi integer (angka)

Menampilkan data pada baris tertentu

```python
# Menampilkan data yang berada di baris ke-33
df.iloc[33]
```

Menampilkan data dengan kondisi tertentu

```python
# Menampilkan data tweet yang memiliki value/nilai lebih dari sama dengan 1200 likes menggunakan .loc[]
df.loc[df["favorite_count"] >= 1200]
```

```python
# Menampilkan data tweet yang memiliki nilai antara 51 hingga 100 likes menggunakan .loc[]
df.loc[(df["favorite_count"] <= 100) & (df["favorite_count"] > 50)]
```

Menampilkan data pada baris dan kolom tertentu

```python
# Menampilkan data pada baris ke-100 dari kolom "username"
df.loc[100, "username"]
```

Menampilkan data pada rentang baris dan kolom tertentu

```python
# Menampilkan data yang berada di kolom "location" di rentang dari baris/index ke 1001-1010
# syntax => df.iloc[rentang baris, kolom ke-(dihitung dari 0 milik "tgl_tweet")]

df.iloc[1001:1011, 10]
```

#### **Filtering Data**

Karena data yang kita punya cukup banyak (>2000 baris), kita akan coba untuk filter data dengan ketentuan:

- Memiliki 100 baris
- Urut dari likes terbanyak

##### Langkah 1: Buat variabel untuk menampung data baru hasil filtering

```python
# Buat variabel untuk menampung data baru hasil filtering.
# Variabel data baru akan berisi copy dari data lama, antisipasi supaya data awal tidak berubah

df1 = df.copy()
df1
```
