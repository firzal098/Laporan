```python
import pandas as pd
from datetime import datetime
```

# **1. MEMBACA DATASET & PERSIAPAN**

```python
# Membaca file dataset Data Alumni.csv
df = pd.read_csv("data_alumni.csv")
df.head()
```

```python
# Mengubah nama kolom agar konsisten dan mudah diakses
df.columns = ['No','NIM','Nama','TTL','Fakultas','Alamat','Masuk','Lulus','Status','MasaTunggu','Gaji','TOEFL','IPK']
df
```

# **2. SELEKSI DATA: SAMPLING (SUBSETTING KOLOM)**

```python
# Mengambil subset kolom tertentu (Nama dan TTL) menggunakan .loc
df.loc[:, ['Nama',"TTL"]]
```

# **3. SELEKSI DATA: FEATURE EXTRACTION (EKSTRAKSI FITUR)**

```python
# Ekstraksi Tahun Lahir dari string TTL (mengambil 4 karakter terakhir)
for index, row in df.iterrows():
    df.loc[index, 'Tahun Lahir'] = row['TTL'][-4:]
df
```

```python
# Perhitungan Umur (Tahun Sekarang - Tahun Lahir)
now = datetime.now()
now.year
for index, row in df.iterrows():
    df.loc[index, 'Umur']=now.year - int (row['Tahun Lahir'])
df
```

```python
# Perhitungan Masa Studi (Tahun Lulus - Tahun Masuk)
for index, row in df.iterrows():
    df.loc[index, 'Masa Studi']=row['Lulus']-row['Masuk']
df
```

# **4. SELEKSI DATA: DENOISING (PEMBERSIHAN DATA SISA EXTRACT)**

```python
# Cek missing value / noise pada fitur-fitur baru yang telah dibuat
print("--- Cek Missing Value pada Fitur Hasil Ekstraksi ---")
print(df[["Masa Studi", 'Umur', "Tahun Lahir"]].isnull().sum())
```

# **5. RINGKASAN HASIL AKHIR**

```python
print("--- Data Akhir Setelah Feature Extraction ---")
df[['Nama', 'TTL', 'Tahun Lahir', 'Umur', 'Masa Studi']].head()
```

```python
# Ringkasan proses seleksi data
ringkasan = {
    'Jumlah baris awal': len(df),
    'Jumlah kolom awal': len(df.columns) - 3,
    'Jumlah kolom setelah ekstraksi': len(df.columns)
}

print("\n--- Ringkasan Proses Seleksi Data ---")
for k, v in ringkasan.items():
    print(f'{k:32s}: {v}')
```
