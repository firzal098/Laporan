tidak ditemukan data yang kosong, maka bisa kita langsung proses ke tahap selanjutnya

```python
# Menghapus kolom No, Alamat, dan Tempat & tanggal lahir karena tidak diperlukan
df = df.drop(columns=['No', 'Alamat', 'Tempat & tanggal lahir'])
df.head()
```

# Transformasi Data

## Kategorisasi Data IPK

```python
'''
Mengubah data IPK berdasarkan kategori tertentu
Jika IPK > 3.5 maka predikat sangat memuaskan
Jika IPK > 3 maka predikat memuaskan
Jika IPK < 3 maka predikat cukup
'''

for index, row in df.iterrows():
  if row['IPK'] > 3.5:
    df.loc[index, 'predikat'] = "Sangat Memuaskan"
  elif row['IPK'] > 3:
    df.loc[index, 'predikat'] = "Memuaskan"
  else:
    df.loc[index, 'predikat'] = "Cukup"

df[["Nim", 'Nama', 'IPK', 'predikat']]
```

## Kategorisasi Data Masa Studi

```python
'''
Mengubah data masa studi berdasarkan kategori tertentu
Jika masa studi > 4 tahun maka kategori masa studi diwaktu yang tepat
Jika masa studi < 4 tahun maka kategori masa studi tepat waktu
'''

for index, row in df.iterrows():
  if row['Masa Studi'] > 4:
    df.loc[index, 'Kategori Masa Studi'] = "Diwaktu yang tepat"
  else:
    df.loc[index, 'Kategori Masa Studi'] = "Tepat Waktu"

df[["Nim", 'Nama', 'Masa Studi', 'Kategori Masa Studi']]
```

# Min-Max Data Nilai TOEFL

```python
'''
Min-max scaler cocok digunakan ketika data memiliki distribusi yang normal
'''

from sklearn.preprocessing import MinMaxScaler

minMax = MinMaxScaler()
```

```python
'''
Mengubah data nilai TOEFL menjadi skala min-max
Nilai TOEFL akan berubah menjadi nilai antara 0 sampai 1
Nilai yang paling rendah akan berubah menjadi 0
Nilai yang paling tinggi akan berubah menjadi 1
Rumus min-max = (nilai - nilai minimum) / (nilai maksimum - nilai minimum)
'''

df['min-max TOEFL'] = minMax.fit_transform(df[["Nilai Toefl"]])

df[['Nim', 'Nama', 'Nilai Toefl', 'min-max TOEFL']]
```

## Standard Scaler Data Gaji

```python
'''
Standard scaler cocok digunakan ketika data memiliki distribusi
yang tidak normal atau memiliki nilai outlier yang signifikan
'''

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
'''
Mengubah data gaji menjadi skala standar
Nilai gaji akan berubah menjadi nilai yang memiliki rata-rata 0 dan deviasi standar 1
Rumus standar = (nilai - rata-rata) / deviasi standar
'''

df['standard Gaji'] = scaler.fit_transform(df[['Gaji Pertama bekerjaa']])

df[['Nim', 'Nama', 'Gaji Pertama bekerjaa', 'standard Gaji']]
```

## One Hot Encoding Predikat

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder()
```

```python
'''
Mengubah data Kategori Masa Studi menjadi data biner
Data Kategori Masa Studi akan berubah menjadi data biner yang terdiri dari 3 kolom
Kolom pertama akan berisi nilai 1 jika Kategori Masa Studi adalah cukup
Kolom kedua akan berisi nilai 1 jika Kategori Masa Studi adalah memuaskan
Kolom ketiga akan berisi nilai 1 jika Kategori Masa Studi adalah sangat memuaskan
'''

encoder_df = pd.DataFrame(encoder.fit_transform(df[['Kategori Masa Studi']]).toarray())

encoder_df.head()
```

```python
# Mengubah nama kolom Kategori Masa Studi setelah dilakukan one hot encoding
new_columns_for_kms = encoder.get_feature_names_out(['Kategori Masa Studi'])

encoder_df.columns = new_columns_for_kms

encoder_df.head()
```

```python
# Menggabungkan data awal dengan data one hot encoding predikat
df = df.join(encoder_df)
```

```python
df.head()
```

## Label Encoder Data Kategori Masa Studi

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
```

```python
'''
Mengubah data predikat menjadi data numerik
Data predikat akan berubah menjadi data numerik yang terdiri dari 3 nilai
Nilai pertama akan berisi nilai 0 jika predikat adalah cukup
Nilai kedua akan berisi nilai 1 jika predikat adalah memuaskan
Nilai ketiga akan berisi nilai 2 jika predikat adalah sangat memuaskan
'''

df['le_predikat'] = le.fit_transform(df['predikat'])
```

```python
df[['Nim', 'Nama', 'predikat','le_predikat']].head()
```
