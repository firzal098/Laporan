```python
mean_usia = df['Usia_Tahun'].mean()
std_usia = df['Usia_Tahun'].std()
df['zscore_usia'] = (df['Usia_Tahun'] - mean_usia) / std_usia
outlier_zscore = df[df['zscore_usia'].abs() > 2]
print(outlier_zscore[['Nama', 'Usia_Tahun', 'zscore_usia']])
```

```python
Q1 = df['Usia_Tahun'].quantile(0.25)
Q3 = df['Usia_Tahun'].quantile(0.75)
IQR = Q3 - Q1
batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR
outlier_iqr = df[(df['Usia_Tahun'] < batas_bawah) | (df['Usia_Tahun'] > batas_atas)]
print('Batas bawah - atas:', batas_bawah, '-', batas_atas)
print(outlier_iqr[['Nama', 'Usia_Tahun']])
```

```python
# Cek ulang baris mana saja yang mengandung missing value
print(df[df.isnull().any(axis=1)])
# Dropna: menghapus baris yang memiliki missing value
df_dropna = df.dropna()
print('Jumlah baris setelah dropna:', len(df_dropna))
# Fillna: mengisi missing value numerik dengan nilai mean kolomnya
df_fillna = df.copy()
df_fillna['Suhu_Tubuh_C'] = df_fillna['Suhu_Tubuh_C'].fillna(df_fillna['Suhu_Tubuh_C'].mean())
df_fillna['Usia_Tahun'] = df_fillna['Usia_Tahun'].fillna(df_fillna['Usia_Tahun'].median())
print(df_fillna.isnull().sum())
```

```python
# Menghapus baris yang terdeteksi outlier (metode IQR)
df_no_outlier = df[(df['Usia_Tahun'] >= batas_bawah) & (df['Usia_Tahun'] <= batas_atas) | (df['Usia_Tahun'].isnull())]
print('Jumlah baris setelah outlier dihapus:', len(df_no_outlier))
# Capping: membatasi nilai ekstrem ke batas bawah/atas IQR
df_capping = df.copy()
df_capping['Usia_Tahun'] = df_capping['Usia_Tahun'].clip(lower=batas_bawah, upper=batas_atas)
print(df_capping.loc[[9, 30], ['Nama', 'Usia_Tahun']])
# Flagging: menandai baris outlier tanpa mengubah/menghapus nilainya
df['flag_outlier_usia'] = np.where(
  (df['Usia_Tahun'] < batas_bawah) | (df['Usia_Tahun'] > batas_atas),
  'Outlier', 'Wajar'
)
print(df[['Nama', 'Usia_Tahun', 'flag_outlier_usia']])
```

```python
ringkasan = {
  'Jumlah baris awal' : len(df),
  'Jumlah kolom' : df.shape[1],
  'Missing value (sebelum)' : df[['Suhu_Tubuh_C',
  'Usia_Tahun']].isnull().sum().sum(),
  'Outlier terdeteksi (IQR)' : len(outlier_iqr),
  'Jumlah baris setelah dropna': len(df_dropna),
}
for k, v in ringkasan.items():
  print(f'{k:28s}: {v}')
```
